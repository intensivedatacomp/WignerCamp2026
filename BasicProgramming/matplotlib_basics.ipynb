## Introduction to `Matplotlib`
`Matplotlib` is a plotting library for the `Python` programming language and its numerical mathematics extension `NumPy`. It provides an object-oriented `API` for embedding plots into applications.

```python
# Importing Matplotlib
import matplotlib.pyplot as plt
```

## Changing Markers, Line Styles, and Colors
You can customize the appearance of your plots by changing markers, line styles, and colors.

```python
# Example plot with different markers, line styles, and colors
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, marker='o', linestyle='-', color='r', label='Line 1')
plt.plot(x, [15, 25, 20, 35], marker='s', linestyle='--', color='b', label='Line 2')
plt.legend()
```

## Setting `LaTeX` Fonts
You can configure the font family, size, and enable `LaTeX` rendering in `Matplotlib` using `plt.rcParams.update()`. This is especially useful for scientific plots where mathematical notation is needed.

```python
# Example: configuring LaTeX fonts
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.size": 25
})

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, marker='o')
plt.title(r'$\sin(x)$ with \LaTeX{} font')
plt.xlabel(r'$x$ (a. u.)')
plt.ylabel(r'$y$ (a. u.)')
```

## Adding Text to Plots
You can add text annotations to your plots using the `text` function. It is good practice to offset the text position slightly from the data point to avoid overlapping with markers.

```python
# Example plot with text annotations
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, marker='o')
plt.text(2.1, 18, 'This is a point')
```

## Titles
You can add titles to your plots using the `title` function. The `pad` argument controls the spacing between the title and the top of the axes in points.

```python
# Example plot with a title
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, marker='o')
plt.title('Sample Plot', pad=10)
```

## Subplots
You can create multiple subplots in a single figure using the `subplots` function.

```python
# Example with multiple subplots
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

# First subplot
x1 = [1, 2, 3, 4]
y1 = [10, 20, 25, 30]
ax[0].plot(x1, y1, marker='o')
ax[0].set_title('Subplot 1')

# Second subplot
x2 = [1, 2, 3, 4]
y2 = [30, 25, 20, 10]
ax[1].plot(x2, y2, marker='s')
ax[1].set_title('Subplot 2')
```
