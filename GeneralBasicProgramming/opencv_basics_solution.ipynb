# OpenCV
OpenCV is an opensurce computer vision libary in `python` and `C++`.

## Image read and write

```python
import cv2
import matplotlib.pyplot as plt

if IN_COLAB:
    img = cv2.imread("/content/drive/My Drive/WignerCamp/img/cat.jpg") 
else:
    img = cv2.imread("../img/cat.jpg") 

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # Convert BGR to RGB

plt.imshow(img_rgb) # Display the image
plt.axis("off") # Turn off axis labels and ticks
plt.show() # Show the plot
```

## Resize the image

### Resize to `280x600´

```python
small = cv2.resize(img, (280, 600))

plt.imshow(cv2.cvtColor(small, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

### Original vs Resized

```python
small = cv2.resize(img, (280, 500))

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(cv2.cvtColor(small, cv2.COLOR_BGR2RGB))
ax[1].set_title("Resized")
ax[1].axis("off")

plt.show()
```

## Crop 

```python
crop = img[100:300, 30:500]

plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

## Convert to gray scale image

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

## Canny Edge Detector
__Canny__ is a classical edge detector algorithm 

```python

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 500, 600)

plt.imshow(edges, cmap="gray")
plt.axis("off")
plt.show()
```
