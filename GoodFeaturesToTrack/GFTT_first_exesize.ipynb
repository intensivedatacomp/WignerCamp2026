# In this notebook make some easy exesize how GoodFeaturesToTrack dose work

```python
import cv2
from matplotlib import pyplot as plt

img = cv2.imread("../img/chair.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

corners = cv2.goodFeaturesToTrack(
    gray,
    maxCorners=100,
    qualityLevel=0.01,
    minDistance=10
)

for c in corners:
    x, y = c.ravel()
    cv2.circle(img, (int(x), int(y)), 7, (0, 255, 0), -1)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```
